## Step4 : Remove Personal Information

```python
# These columns contain personal contact information.
# They are not needed for hotel booking analysis.

personal_columns = [
    "name",
    "email",
    "phone-number",
    "credit_card"
]

data1.drop(
    columns=personal_columns,
    errors="ignore",
    inplace=True)
data1.shape
```

## Step6 : Data Cleaning

```python
# Remove extra spaces from column names
data1.columns.str.strip()
```

```python
data1.replace(np.nan,'?' , inplace = True)
```

```python
# Convert numeric columns into numeric data types

numeric_columns = [
    "is_canceled",
    "lead_time",
    "arrival_date_year",
    "arrival_date_week_number",
    "arrival_date_day_of_month",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies",
    "is_repeated_guest",
    "previous_cancellations",
    "previous_bookings_not_canceled",
    "booking_changes",
    "agent",
    "company",
    "days_in_waiting_list",
    "adr",
    "required_car_parking_spaces",
    "total_of_special_requests"
]

for col in numeric_columns:

    if col in data1.columns:

        data1[col] = pd.to_numeric(
            data1[col],
            errors="coerce"
        )
```

```python
# Fill missing numeric values using median

for col in numeric_columns:

    if col in data1.columns and data1[col].isnull().any():

        data1[col] = data1[col].fillna(
            data1[col].median()
        )
```

```python
# Fill missing categorical values with Unknown

categorical_columns = data1.select_dtypes(
    include="object"
).columns

for col in categorical_columns:

    data1[col] = data1[col].fillna("Unknown")
```

```python
# Fill missing reservation dates with the mode date

if data1["reservation_status_date"].isnull().any():

    mode_date = data1["reservation_status_date"].mode()

    if not mode_date.empty:

        data1["reservation_status_date"] = (
            data1["reservation_status_date"].fillna(mode_date[0])
        )
```

## Step8 : Data Cleaning

```python
# Make a copy of original dataset
df_clean = df.copy()

# Remove extra spaces from column names
df_clean.columns = df_clean.columns.str.strip()

# Remove extra spaces from text values
for column in df_clean.select_dtypes(include=["object", "str"]).columns:
    df_clean[column] = df_clean[column].str.strip()

# Replace ? and empty strings with NaN
df_clean.replace(["?", ""], np.nan, inplace=True)

print("\nMissing Values After Replacing ?:")
print(df_clean.isnull().sum())
```

## Step9 : Remove Duplicates

```python
before_duplicates = df_clean.shape[0]
df_clean.drop_duplicates(inplace=True)
after_duplicates = df_clean.shape[0]

print("\nDuplicate Rows Removed:")
print(before_duplicates - after_duplicates)

print("Dataset Shape After Removing Duplicates:")
print(df_clean.shape)
```

## Step10 : Convert Numeric Columns

```python
numeric_columns = [
    "age",
    "fnlwgt",
    "educational-num",
    "capital-gain",
    "capital-loss",
    "hours-per-week"]
for column in numeric_columns:
    df_clean[column] = pd.to_numeric(
        df_clean[column], errors="coerce")
print("\nNumeric Columns Converted Successfully!")
```

## Step11 : Handle Missing Values

```python
# Fill numeric missing values with median
for column in numeric_columns:
    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median())
    
# Select categorical columns
categorical_columns = df_clean.select_dtypes(
    include=["object", "str"]).columns

# Fill categorical missing values with Unknown
for column in categorical_columns:
    df_clean[column] = df_clean[column].fillna("Unknown")

# Display missing values after cleaning
print("\nMissing Values After Cleaning:")
print(df_clean.isnull().sum())

# Display total missing values
print("\nTotal Missing Values After Cleaning:")
print(df_clean.isnull().sum().sum())
```

```python
df.dtypes
```

```python
# Convert numeric columns

numeric_columns = [
    "ApplicantIncome",
    "CoapplicantIncome",
    "LoanAmount",
    "Loan_Amount_Term",
    "Credit_History"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Fill missing numeric values using median

for col in numeric_columns:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].median())
```

```python
# Fill missing categorical values using mode

categorical_columns = df.select_dtypes(
    include="object"
).columns.tolist()
for col in categorical_columns:
    if col == "Loan_Status":
        continue
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Remove rows with missing target values

df.dropna(subset=["Loan_Status"], inplace=True)
```
